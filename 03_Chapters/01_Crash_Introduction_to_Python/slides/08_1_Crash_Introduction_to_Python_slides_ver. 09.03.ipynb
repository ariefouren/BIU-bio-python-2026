## 1.1. A Crash Example: GC content

It demonstrates:

- <span class="fragment">function definition</span>
- <span class="fragment">variables</span>
- <span class="fragment">a loop over a sequence</span>
- <span class="fragment">a conditional statement</span>
- <span class="fragment">a return value</span>


```python
def GC_content(dna_string):
    ''' calculate %GC in dna_string (uppercase ACTG) '''
    GC = 0
    for nuc in dna_string: # iterate over dna_string
        if nuc == "C" or nuc == "G":
            GC += 1 # equal to GC = GC+1
    return 100*GC/len(dna_string)
```

```python
print(GC_content.__doc__)
```

```python
GC_content("GT")
```

```python
GC_content("CCG")
```

```python
GC_content("TTAGACCAGTAGAAGTAC")
```

## Main keypoints:
### Structure of the function
- `def` starts a function definition
- indentation defines the function body
- the docstring describes the function

### Main logic
- scan the DNA string one nucleotide at a time
- count only `C` and `G`
- divide by total length
- multiply by 100 to get a percentage

### Self-checking questions
- What happens on an empty string?
- What happens on lowercase input?
- Could we write the same idea in a different way?


### Visualization: nucleotide histogram


```python
from collections import Counter
import matplotlib.pyplot as plt

sample_dna = "TTAGACCAGTAGAAGTAC"
counts = Counter(sample_dna)

bases = ["A", "C", "G", "T"]
values = [counts.get(base, 0) for base in bases]

plt.figure(figsize=(7, 4))
plt.bar(bases, values, color=["blue", "orange", "green", "red"])
plt.title("Nucleotide counts in a sample DNA sequence")
plt.xlabel("Base")
plt.ylabel("Count")
plt.show()
```

# 1.6. Functions
Functions are reusable pieces of code that perform a specific task.
- Built-in functions: `print()`, `len()`, `int()`, `str()`, etc. Can be used without definition.
- User-defined functions: defined by the programmer using `def` keyword.

## 1.6.1. Built-in functions 
Build-in functions are part of the Python language and are always available. For example: `len(...)`, `max(...)`, `min(...)`, `sum(...)`, `sorted(...)`, `range(...)`.


```python
values = [1, 2, 3, 4, 5]
print("min:", min(values))
print("max:", max(values))
print("sum:", sum(values))
```

```python
# min and max also work with strings, 
# based on lexicographical order (dictionary order)
print("min: ", min(["cat", "dog", "bee"]))
print("max: ", max(["cat", "dog", "bee"]))
```

## Built-in function `sorted(...)`
Function `sorted(...)` returns a sorted version of a sequence (string or list), without modifying the original sequence.

```python
values = [ 10, -1, 7, -12, 18, 0]
print("sorted values:", sorted(values))
print("original values:", values) # note that the original list is unchanged
```

## 1.6.2. User-defined functions
User-defined functions are defined by the programmer to perform specific tasks. They are created using the `def` keyword, followed by the function name and parameters.


```python
# Function definition:
def CG_content(dna_string):
    ''' calculate %GC in dna_string (uppercase ACTG) ''' # docstring: a string that describes the function
    # function body: the code that performs the calculation
    # note the indentation ( <tab> ) of the function body
    GC = 0
    for nuc in dna_string: # iterate over dna_string
        if nuc == "C" or nuc == "G":
            GC += 1 # equal to GC = GC+1
    return 100*GC/len(dna_string)   # return value

# Function call:
print(CG_content("GT"))

# Function return value can be assigned to a variable for later use:
result = CG_content("ATGCGTAAGGDTAGTAAGGTAAGGTAAGGC")
print(result)
```

## 1.6.3. Class Methods
In addition to the built-in and user-defined function, additional sources for functions are class
methods. These are functions that belong to specific classes in Python. For example, there are
methods that belong to the class of strings.

```python
# Count the number of occurrences of "G" and "A" 
dna = "ATGCGTAAGGDTAGTAAGGTAAGGTAAGGC"
print('''dna.count("G") :''', dna.count("G")) # built-in string method to count the number of occurrences of "G" in the string
print('''dna.count("A") :''', dna.count("A"))
```

```python
# String methods do not modify the original string, 
# they return a new string
print('''dna.lower() :          ''', dna.lower()) # built-in string method to convert the string to lowercase
print('''dna remains unchanged: ''', dna) # note that the original string is unchanged
```

```python
# Replace all occurrences of "A" with "-" in the string
# Note that this does not modify the original string, 
# it returns a new string with the replacements
print('''dna.replace("A", "-") :''', dna.replace("A", "-")) # built-in string method to replace all occurrences of "D" with "A" in the string
print('''dna remains unchanged: ''', dna) # note that the original string is unchanged
```

```python
# Find the index of the first occurrence of a substring in the string
print('''dna :            ''', dna)
print('''dna.find("G") :  ''', dna.find("G")) # built-in string method to find the index of the first occurrence of "A" in the string
print('''dna.find("TAA") :''', dna.find("TAA"))
print('''dna.find("X") :  ''', dna.find("X")) # returns -1 if the substring is not found in the string
```

## Conditional statements

Conditionals split the program into alternative execution paths.

A very clean example from the chapter:


```python
def state(Tm, Tv, T):
    ''' return the state of material with Tm and Tv at temperature T '''
    if T < Tm:
        return "solid"
    else:
        if T < Tv:
            return "liquid"
        else:
            return "gas"
```

```python
state(0, 100, 25)
```

### Main keypoints
- `if` checks a condition
- `else` gives the alternative branch
- indentation determines which statements belong to each branch
- this is a natural place to ask students to predict outputs before running the cell


## Iteration with `for` loops

The chapter introduces `for` loops as a natural way to iterate over strings and lists.


```python
def basic_aa_perc(prot):
    ''' return percentage of basic amino acids in protein '''
    basic = 0
    basic_aa_list = ["K", "H", "R"]
    for aa in prot:
        if aa in basic_aa_list:
            basic += 1
    return 100*basic/len(prot)
```

```python
basic_aa_perc("MHKRAGG")
```

## Iteration with `while` loops

A `while` loop repeats as long as a condition remains true.

This version rewrites GC content with an explicit running index.


```python
def GC_content3(dna_string):
    ''' calculate %GC in dna string (uppercase ACTG) '''
    GC = 0
    i = 0
    while i < len(dna_string):
        if dna_string[i] == "C" or dna_string [i] == "G":
            GC += 1
        i += 1
    return 100*GC/len(dna_string)
```

```python
GC_content3("TTAGACCAGTAGAAGTAC")
```
