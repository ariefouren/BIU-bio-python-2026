## 1.1. A Crash Example: GC content

It demonstrates:
- function definition
- variables
- a loop over a sequence
- a conditional statement
- a return value

````python
def GC_content(dna_string):
    ''' calculate %GC in dna_string (uppercase ACTG) '''
    GC = 0
    for nuc in dna_string: # iterate over dna_string
        if nuc == "C" or nuc == "G":
            GC += 1 # equal to GC = GC+1
    return 100*GC/len(dna_string)
````

### Comparison operators

| Operator | Symbol | Example |
|---|---|---|

| Less than | `<` | `x < y` |
| Less or equal | `<=` | `x <= y` |

**The rules**
- A comparison gives a `bool`: `True` or `False`
- `int` and `float` compare by value: `3 == 3.0` → `True`
- Arithmetic is computed before comparison: `2 + 2 == 4` → `True`
- Comparisons can be chained: `1 < 5 < 10` → `True`

## `=` is not `==`

| | `=` assignment | `==` comparison |
|---|---|---|
| Meaning | **store** a value in a variable | **ask**: are the two values equal? |
| Result | nothing is printed; the variable changes | `True` or `False` |
| Example | `x = 5` | `x == 5` |

**Common errors**

1. `=` inside a condition — Python stops with an error:
   ```python
   if x = 5:
   # SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?
   ```
2. `==` instead of `=` — **no error**, but nothing is stored (a silent bug):
   ```python
   x = 3
   x == 5      # only compares (False) — x is still 3
   ```
3. Comparing values of different types — no error, just `False`: `"5" == 5` is `False`.
4. Decimal numbers are stored approximately: `0.1 + 0.2 == 0.3` is `False`.

## `=` is not `==`
### What will be the output?

````python
x = 3
x == 5          # a comparison, not an assignment
print("x =", x)
print('"5" == 5 :', "5" == 5)
print("0.1 + 0.2 == 0.3 :", 0.1 + 0.2 == 0.3)
print("0.1 + 0.2 =", 0.1 + 0.2)
````

## Answer
Why: x == 5 only compares; "5" is a str; decimals are stored approximately

````python
x = 3
x == 5          # a comparison, not an assignment
print("x =", x)
print('"5" == 5 :', "5" == 5)
print("0.1 + 0.2 == 0.3 :", 0.1 + 0.2 == 0.3)
print("0.1 + 0.2 =", 0.1 + 0.2)
````

## Strings are comparable
Python compares strings lexicographically (like in a dictionary).

**The rules**
- Compare **character by character** from the left; the **first difference** decides.
- Each character has a numeric code (Unicode; for English letters and digits it is the same as ASCII).
  `ord("A")` gives the code, `chr(65)` gives the character.
- Order of codes: **space < digits < CAPITAL letters < small letters**
- If one string is the **beginning** of the other, the **shorter** one is smaller: `"AA" < "AAA"`
- Digits are compared as characters, not as numbers: `"10" < "9"` is `True` (`"1"` < `"9"`)
- A number cannot be compared with a string: `5 < "A"` raises a `TypeError`
- To ignore capital/small letters, compare `s.lower()`

| Characters | ASCII codes |
|---|---|
| space | 32 |
| `0` … `9` | 48 … 57 |
| `A` … `Z` | 65 … 90 |
| `a` … `z` | 97 … 122 |

Full table: [asciitable.com](https://www.asciitable.com)

## Assignment operator

**The rules**
- `x = 5` means "store 5 in x" — it is **not** a test of equality (that is `==`).
- Python first computes the **right side**, then stores the result in the variable on the left:
  `x = x + 1` takes the old value of `x`, adds 1, and stores the new value back in `x`.
- Short forms (*augmented assignment*):

| Short form | Same as |
|---|---|
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x /= 3` | `x = x / 3` (the result is always a `float`) |
| `x //= 3` | `x = x // 3` |
| `x %= 3` | `x = x % 3` |
| `x **= 3` | `x = x ** 3` |

- The variable must already have a value: `x += 1` on a new variable raises a `NameError`.
- `+=` and `*=` also work on strings and lists: `seq += "CTGA"`, `motif *= 2`.

## Slicing
Use `[i:j]` to get a slice of a sequence from index `i` (inclusive) to `j` (exclusive).
The resulting slice contains elements with indices `i, i+1, ..., j-1`.

**The rules** — `s[start:stop:step]`
- `start` is included, `stop` is **not**: `"ACGTAC"[1:3]` → `"CG"`

- `step` jumps: `"ACGTAC"[::3]` → `"AT"`; a negative step goes backwards
- A slice never raises an error: `"ACGT"[2:100]` → `"GT"`

## 1.6.2. User-defined functions
User-defined functions are defined by the programmer to perform specific tasks. They are created using the `def` keyword, followed by the function name and parameters.

**The rules**
```python
def square(n):        # def, name, parameters, colon
    return n * n      # the body is indented

print(square(4))      # call: 16
```
- A function must be defined **before** it is called
- `return` sends a value back to the caller

````python
# Function definition:
def CG_content(dna_string):
    ''' calculate %GC in dna_string (uppercase ACTG) ''' # docstring
    # function body: the code that performs the calculation
    # note the indentation ( <tab> ) of the function body
    GC = 0
    for nuc in dna_string: # iterate over dna_string
        if nuc == "C" or nuc == "G":
            GC += 1 # equal to GC = GC+1
    return 100*GC/len(dna_string)   # return value

# Function call:
print(CG_content("GT"))

# Function return value can be assigned to a variable for later use:
result = CG_content("ATGCGTAAGGATAGTAAGGTAAGGTAAGGC")
print(result)
````

## 1.9.5 List comprehension
A short way to build a **new list** from a sequence, in **one line**.

````python
# with a for loop:
squares = []
for n in range(1, 6):
    squares.append(n * n)
print(squares)
# the same, with a list comprehension:
squares = [n * n for n in range(1, 6)]
print(squares)
````

## List comprehension: the rules
`[expression for item in sequence]`  or  `[expression for item in sequence if condition]`
- `for item in sequence` takes the elements one by one, like a `for` loop
- `expression` is what goes into the new list for each element: `[x * 2 for x in [1, 2, 3]]` → `[2, 4, 6]`
- `if condition` (optional) keeps only the elements for which it is `True`: `[x for x in [1, 5, 8] if x > 4]` → `[5, 8]`
- It works on any sequence, also on strings: `[b for b in "ACG"]` → `['A', 'C', 'G']`
- The result is always a **new list**; the original sequence does not change

# 1.10 Containers
A container is a data structure that can hold multiple values. Examples of containers in Python include lists, tuples, sets, and dictionaries.

Two important properties of containers:
- **Mutability**: whether the container can be modified after it is created (e.g., lists are mutable, while tuples are immutable).
- **Order**: whether the container maintains the order of elements (e.g., lists and tuples are ordered, while sets and dictionaries are unordered).

| Mutability | Ordered (Sequences)                | Unordered                          |
|------------|------------------------------------|-------------------------------------|
| **Mutable**   | list: `[1, 2, "hello"]`             | set: `{1, 2, "hello"}`<br>dict: `{1:"a", 2:"b", 3:"c"}` |
| **Immutable** | str: `"TGAAC"`<br>tuple: `(1, 2, "hello")` | *(no built‑in examples)*            |

## Sets
A set is an **unordered** collection of **unique** elements. 

Operations on sets include:
- `add(...)`: adds an element to the set
- `remove(...)`: removes an element from the set, and raises an error if the element is not present
- `discard(...)`: removes an element from the set if it is present, and does nothing if it is not present
- `in`: checks if an element is in the set
- `len(...)`: returns the number of elements in the set
- `intersection(...)`: returns a new set that is the intersection of two sets
- `union(...)`: returns a new set that is the union of two sets

To create a set, we can use curly braces `{}` or the `set(...)` constructor.

**The rules**
- Each element appears **once**: `{1, 1, 2}` → `{1, 2}`
- No order and no indexing: `s[0]` raises a `TypeError`
- `{}` is an empty **dictionary** — an empty set is `set()`

## Dictionaries
A dictionary is an **unordered** collection of key-value pairs. Each key is unique and maps to a value.

Dictionaries are used to map keys to values, and they provide efficient lookup based on keys. 

To get the value associated with a key, we can use `dict[key]`. 

To add or update a key-value pair, we can use `dict[key] = value`.

**The rules**
- `d = {"Lys": "K"}` — keys are unique; `d["Lys"]` → `"K"`
- A missing key raises a `KeyError`; `d.get("Met", "?")` returns `"?"` instead

## Mapping DNA codons to the amino acids using a dictionary

````python
# mapping of codons to amino acids 
# according to the standard genetic code
# `*` represents stop codons
universal = {
    'ttt': 'F', 'tct': 'S', 'tat': 'Y', 'tgt': 'C',
    'ttc': 'F', 'tcc': 'S', 'tac': 'Y', 'tgc': 'C',
    'tta': 'L', 'tca': 'S', 'taa': '*', 'tga': '*',
    'ttg': 'L', 'tcg': 'S', 'tag': '*', 'tgg': 'W',

    'ctt': 'L', 'cct': 'P', 'cat': 'H', 'cgt': 'R',
    'ctc': 'L', 'ccc': 'P', 'cac': 'H', 'cgc': 'R',
    'cta': 'L', 'cca': 'P', 'caa': 'Q', 'cga': 'R',
    'ctg': 'L', 'ccg': 'P', 'cag': 'Q', 'cgg': 'R',

    'att': 'I', 'act': 'T', 'aat': 'N', 'agt': 'S',
    'atc': 'I', 'acc': 'T', 'aac': 'N', 'agc': 'S',
    'ata': 'I', 'aca': 'T', 'aaa': 'K', 'aga': 'R',
    'atg': 'M', 'acg': 'T', 'aag': 'K', 'agg': 'R',

    'gtt': 'V', 'gct': 'A', 'gat': 'D', 'ggt': 'G',
    'gtc': 'V', 'gcc': 'A', 'gac': 'D', 'ggc': 'G',
    'gta': 'V', 'gca': 'A', 'gaa': 'E', 'gga': 'G',
    'gtg': 'V', 'gcg': 'A', 'gag': 'E', 'ggg': 'G'
}
````

## Mapping DNA codons to the amino acids using a dictionary
### What will be the output?

````python
def dna_translate(gene, code=universal):    # default: universal code
    prot = ""
    gene = gene.lower()          # the dictionary has lowercase codons
    for i in range(0, len(gene), 3):
        codon = gene[i:i+3]
        prot += code.get(codon, "?")     # "?" = unknown codon
    return prot
print(dna_translate("ATGGACGGCGGCTAA"))
````

## Answer
Why: Codons ATG GAC GGC GGC TAA → M D G G `*` (`*` is a stop codon).

## Tuples

Tuples are immutable ordered sequences. They are similar to lists, but cannot be modified after creation.

Tuples are defined using parentheses `(...)` and can contain elements of any type.

### 📁 What Are Files?
- Used to store **long‑term data** beyond a program’s execution.

### 💾 Why Files Matter
- Can hold **much larger amounts of data** than variables in memory.
- Data **persists** even after:
  - the program ends  
  - the computer shuts down  

### 🧬 Files in Biology
- Essential for storing:
  - **genomes** of organisms  
  - **gene expression data** from healthy vs. mutant cells  

### 🖥️ How Files Are Used
- Programs later **read** these files to perform analyses and extract insights.

## Working with Files in Python
### 📂 Opening a File
- Use `open(filename, mode)` to open a file.
- `filename`: the name of the file (e.g., `"data.txt"`)
- `mode`: how you want to interact with the file:
  - `'r'`: read (default)
  - `'w'`: write (creates a new file or overwrites an existing one)
  - `'a'`: append (adds to the end of an existing file)

### 📖 Reading from a File
- Use `file.read()` to read the entire contents of the file as a string.
- Use `file.readline()` to read one line at a time.
- Use `file.readlines()` to read all lines into a list.

### ✍️ Writing to a File
- Use `file.write(string)` to write a string to the file.
- Use `file.writelines(list_of_strings)` to write a list of strings to the file.

### 🔒 Closing a File
- Always close a file after you're done using it with `file.close()`.
- Or use `with open(filename, mode) as f:` — the file is closed automatically at the end of the block.


## Writing and reading a small file
`with open(...) as f:` opens the file, and **closes it automatically** at the end of the block.

````python
with open("dna.txt", "w") as f:        # "w": create a new file (or overwrite)
    f.write("ATGGCCATTGTA\n")
with open("dna.txt", "a") as f:        # "a": add at the end of the file
    f.write("TGCCGA\n")
with open("dna.txt") as f:             # "r" (the default): read
    print(f.read())
````

````python
# Example: generate a random DNA sequence of length 50
import random # import the random module to generate random numbers
def random_dna(length):
    ''' generate a random DNA sequence of given length '''
    bases = ["A", "C", "G", "T"] # list of DNA bases
    dna = ""
    for i in range(length):
        # randomly select a base and append to the dna string
        dna += random.choice(bases)
    return dna

print(random_dna(50)) # generate and print a random DNA sequence of length 50
# generate and print another random DNA sequence of length 50
print(random_dna(50))
````

## Python

````python
# Example: generate a random permutation of numbers from 1 to n
def random_permutation(n):
    ''' generate a random permutation of numbers from 1 to n '''
    numbers = list(range(1, n+1)) # create a list of numbers from 1 to n
    # shuffle the list in place to get a random permutation
    random.shuffle(numbers)
    return numbers

# generate and print a random permutation of numbers from 1 to 10
print(random_permutation(10))
# generate and print another random permutation of numbers from 1 to 10
print(random_permutation(10))
````
