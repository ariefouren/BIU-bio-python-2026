## 6.1 The Shortest Paths Problem

Given a graph and two nodes, `v1` and `v2`, we want to find:

- the distance `dist(v1, v2)`
- a path whose length equals this distance

That is, we want a **shortest path** from `v1` to `v2`.

## Applications of shortest paths

Shortest path algorithms appear in many contexts:

- navigation instructions
- routing information through a network
- measuring distance in a social graph
- finding signaling paths in biological networks

In this chapter, the graph is **non-weighted**: the length of a path is the number of edges.

## 6.2 Breadth First Search (BFS)

BFS is a graph traversal algorithm.

Starting from a source node, it discovers nodes in waves:

- distance 0: the source node
- distance 1: immediate neighbors
- distance 2: neighbors of neighbors
- and so on

At the end, every reachable node has its correct distance from the source.

## 6.2.1 The Algorithm

BFS maintains a list of nodes waiting to be handled.

These nodes are called **active nodes**.

Initially:

- only the source node is active
- all other nodes are unseen

When an unseen node is discovered, it enters the list of active nodes.

## FIFO queue

BFS uses a queue.

FIFO means:

**First In, First Out**

The node that has waited the longest is handled first.

## BFS pseudo-code 

```text
BFS(G, source):

Initialization:

1. dists = 0 for source, and infinity for all other nodes in G
2. Q = [source], the only node we “discovered” so far
```

```text
Main Loop:

3. while Q is not empty:
   3.1 u ← the head of Q (oldest node)
   3.2 remove u from Q
   3.3 for every neighbor v of u in G:
       3.3.1 if v is seen for the first time:
             3.3.1.1 update dist[v] to be dist[u] + 1
             3.3.1.2 insert v into Q
```

```text
Termination (Q is empty):

4. return dists
```

## BFS pseudo-code — Termination

```text
Termination (Q is empty):

4. return dists
```

## BFS implementation

The function `bfs(G, source)` receives:

- `G`: a graph represented as a binary adjacency matrix
- `source`: the starting node

It returns:

- `dists`: a list of distances from `source`

````python
import math                 #for math.inf

def bfs(G, source):
    '''run breadth first search on graph G, starting from source node
    G is represented as an adjacency matrix, where G[u][v] == 1 if there
    is an edge from u to v, and G[u][v] == 0 otherwise.
    returns a list of distances from source to each node, where distance is
    the number of edges in the shortest path from source to that node, or
    math.inf if there is no path from source to that node.'''
    
    inf = math.inf
    n = len(G) # number of rows in matrix = number of nodes

    # Initialization
    dists = [inf]*n     # create list of n elements, all set to inf
    dists[source] = 0   # distance from source to itself is 0, 
                        # all other distances are inf

    Q = [source] # Queue of "active" nodes, initially only source is active

    # Main loop of the algorithm
    while len(Q) != 0: # while there are still nodes to handle
        u = Q.pop(0) # get and remove head of Q
        for v in range(n): # go over all neighbors v of u
            if G[u][v] == 1 and dists[v] == inf: # first time in v
                dists[v] = dists[u] + 1 # update v's dist
                Q.append(v) # v enters Q

    # Termination, Q is empty
    return dists
````

## Code explained

- `inf` marks nodes not reached yet.
- `len(G)` is the number of rows in the adjacency matrix, hence the number of nodes.
- `Q.pop(0)` removes the oldest active node.
- `G[u][v] == 1` means that there is an edge from `u` to `v`.
- When `v` is seen for the first time, its distance becomes `dists[u] + 1`.

## 6.2.3 Reconstructing the Shortest Paths

So far, BFS computed only the **lengths** of the shortest paths.

Now we want to reconstruct the paths themselves.

Idea:

- Store, for each node, the node that discovered it.
- This list is called `prevs`.
- Then reconstruct the path backwards from target to source.

## Modified BFS with previous nodes

The modified BFS returns two lists:

- `dists`: distances from the source
- `prevs`: previous node on the shortest path

````python
def bfs(G, source):
    inf = math.inf
    n = len(G) # number of rows in matrix = number of nodes

    # Initialization
    dists = [inf]*n
    dists[source] = 0
    prevs = [-1]*n # initial values of -1 denote node not discovered yet

    Q = [source] # Queue of "active" nodes

    # Main loop of the algorithm
    while len(Q) != 0: # while there are still nodes to handle
        u = Q.pop(0) # get and remove head of Q
        for v in range(n): # go over all neighbors v of u
            if G[u][v] == 1 and dists[v] == inf: # v seen first time
                dists[v] = dists[u] + 1 # update v's dist
                prevs[v] = u # u is v's previous
                Q.append(v) # v enters Q

    # Termination, Q is empty
    return dists, prevs
````

## The function `shortest_path`

The function receives:

- `G`: a graph
- `source`: source node
- `target`: target node

It returns the shortest path from `source` to `target`.

````python
def shortest_path(G, source, target):
    ''' Returns the shortest path in G from source to target.
    Uses BFS's output.
    G is non-weighted, directed graph as a binary matrix.
    The output is a list of the nodes in the path '''
    dists, prevs = bfs(G, source) # dists will not be used

    if prevs[target] == -1: # target not reachable from source
        return []

    path = [target]
    v = target

    while v != source:
        v = prevs[v] # previous node in the path
        path = [v] + path # append v at left end of path

    return path
````

## 6.3 Related Biologically Relevant Metrics

Shortest paths help define useful network metrics.

One example is **betweenness**:

the number of shortest paths that go through a node.

A node with high betweenness may play a central role in information transfer through the network.

## Characteristic path length

The **characteristic path length** is:

the average length of all shortest paths between pairs of nodes.

It gives a sense of how “spread out” the graph is.

## Diameter

The **diameter** of a graph is:

the maximal length over all shortest paths.

Equivalently:

the distance between the farthest pair of nodes.

## Exercise 6

Write two functions:

```python
avg_dist(G)
diameter(G)
```

Each receives an unweighted graph `G`, represented as a binary matrix.

They should return:

- the average shortest-path length
- the maximal shortest-path length

````python
def avg_dist(G):
    ''' Return the average finite distance between pairs of nodes in G. '''
    total = 0
    count = 0

    for source in range(len(G)):
        dists, prevs = bfs(G, source)
        for d in dists:
            if d != 0 and d != inf:
                total += d
                count += 1

    if count == 0:
        return 0

    return total / count


def diameter(G):
    ''' Return the maximal finite shortest-path distance in G. '''
    max_dist = 0

    for source in range(len(G)):
        dists, prevs = bfs(G, source)
        for d in dists:
            if d != inf and d > max_dist:
                max_dist = d

    return max_dist
````

## Reflection

BFS solves the shortest paths problem for **unweighted graphs**.

For weighted graphs, BFS is not enough.

A well-known algorithm for non-negative weighted graphs is **Dijkstra's algorithm**.

## BFS and DFS

BFS explores the graph in layers.

DFS explores paths as far as possible before backtracking.

Both are fundamental graph traversal algorithms.
