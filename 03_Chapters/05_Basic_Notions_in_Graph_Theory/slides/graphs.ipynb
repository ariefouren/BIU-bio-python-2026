```python
import random

def rand_graph(n, p):
    G = []
    for i in range(n):
        G.append([0] * n)
        
    for i in range(n):
        for j in range(i+1, n):
            r = random.random()
            if r < p:
                G[i][j] = 1
                G[j][i] = 1

    return G

G = rand_graph(5, 0.15)

for row in G:
    print(row)
```
