## 1.1. A Crash Example: GC content

It demonstrates:
- function definition
- variables
- a loop over a sequence
- a conditional statement
- a return value

````python
def GC_content(dna_string):
    ''' calculate %GC in dna_string (uppercase ACTG) '''
    GC = 0
    for nuc in dna_string: # iterate over dna_string
        if nuc == "C" or nuc == "G":
            GC += 1 # equal to GC = GC+1
    return 100*GC/len(dna_string)
````

````python
print(GC_content.__doc__)
````

````python
GC_content("GT")
````

````python
GC_content("CCG")
````

````python
GC_content("TTAGACCAGTAGAAGTAC")
````

## Explanation of the code
### Structure of the function
- `def` starts a function definition
- indentation defines the function body
- the docstring describes the function

### Main logic
- scan the DNA string one nucleotide at a time
- count only `C` and `G`
- divide by total length
- multiply by 100 to get a percentage

User-defined functions are defined by the programmer to perform specific tasks. They are created using the `def` keyword, followed by the function name and parameters.


````python
# Function definition:
def CG_content(dna_string):
    ''' calculate %GC in dna_string (uppercase ACTG) ''' # docstring: a string that describes the function
    # function body: the code that performs the calculation
    # note the indentation ( <tab> ) of the function body
    GC = 0
    for nuc in dna_string: # iterate over dna_string
        if nuc == "C" or nuc == "G":
            GC += 1 # equal to GC = GC+1
    return 100*GC/len(dna_string)   # return value

# Function call:
print(CG_content("GT"))

# Function return value can be assigned to a variable for later use:
result = CG_content("ATGCGTAAGGDTAGTAAGGTAAGGTAAGGC")
print(result)
````

# 1.7. Conditional statements

Conditionals split the program into alternative execution paths.

Example:


````python
def state(Tm, Tb, T):
    ''' return the state of material with 
    melting point Tm and boiling point Tb
    at temperature T of the environment '''
    if T < Tm:
        return "solid"
    else:
        if T < Tb:
            return "liquid"
        else:
            return "gas"

print(state(0, 100, -10)) # solid
print(state(0, 100, 50)) # liquid
print(state(0, 100, 150)) # gas
````

Another way to write the same logic is using `elif`:

````python
def state(Tm, Tb, T):
    ''' return the state of material with 
    melting point Tm and boiling point Tb
    at temperature T of the environment '''
    if T < Tm:
        return "solid"
    elif T < Tb:
        return "liquid"
    else:
        return "gas"

print(state(0, 100, -10)) # solid
print(state(0, 100, 50)) # liquid
print(state(0, 100, 150)) # gas
````

The chapter introduces `for` loops as a natural way to iterate over strings and lists.


````python
def basic_aa_perc(prot):
    ''' return percentage of basic amino acids in protein '''
    basic = 0
    basic_aa_list = ["K", "H", "R"]
    for aa in prot:
        if aa in basic_aa_list:
            basic += 1
    return 100*basic/len(prot)


basic_aa_perc("MHKRAGG")
````

## Iteration with `while` loops

A `while` loop repeats as long as a condition remains true.

This version rewrites GC content with an explicit running index.


````python
def GC_content3(dna_string):
    ''' calculate %GC in dna string 
        (upper- or lower-case ACTG) '''
    GC_nucleotides = ["C", "G", "c", "g"]
    i = 0
    GC_count = 0
    while i < len(dna_string):
        if dna_string[i] in GC_nucleotides:
            GC_count += 1
        i += 1
    return 100*GC_count/len(dna_string)

print(GC_content3("TTAGACCAGTAGAAGTAC"))
print(GC_content3("ttagaccagtagaagtac"))
````

## `while` loops 
Using `while` loop to find all positions in a given protein where a specific restriction site of an enzyme occurs

````python
def cut_positions(seq, pattern):
    ''' find all positions of pattern within seq '''
    cut_sites = [] # list to store the cut sites
    site = seq.find(pattern) # get index of pattern in seq, -1 if none
    while site != -1:
        cut_sites.append(site)
        site = seq.find(pattern, site+1) # get next match
    return cut_sites

# Example usage:
seq = "AGCTAGCTAGCTAGCT"
pattern = "AGC"
print(cut_positions(seq, pattern)) # should return [0, 4, 8, 12]
````

## Iterating using `for` and `range`
Using `range` to find the first stop codon in
a given DNA sequence, taking into consideration the frame shift

````python
# find the first stop codon in a given DNA sequence, 
# taking into consideration the frame shift
def find_stop(dna):
    stops = ["TAG", "TAA", "TGA"] # list of stop codons
    n = len(dna) 
    for i in range(0, n, 3): # iterate over the dna sequence in steps of 3, to consider the frame shift
        if dna[i:i+3] in stops:
            return i
        
# Example usage:
dna = "ATGCGTAAAGCTAGCTAGCTAGCTAGCTAGC"
print(find_stop(dna)) 
````

## 1.8.4 Iterating over Sequences – Summary
Three ways we have in Python to iterate over a given sequence (string or a list):
 
 1. Using `for` loop directly on the sequence:
 ```python
    for element in sequence:
        # do something with element
```

2. Using `while` loop with an explicit index:

```python
    i = 0
```

```python
    for i in range(len(sequence)):
```

````python
def state(Tm, Tb, T):
    ''' return the state of material with 
    melting point Tm and boiling point Tb
    at temperature T of the environment '''
    if T < Tm:
        return "solid"
    elif T < Tb:
        return "liquid"
    else:
        return "gas"
    

# get user input for Tm, Tb, and T to determine the state of a material
Tm = float(input("Enter the melting point (Tm) of the material: ")) # get user input for melting point and convert to float
Tb = float(input("Enter the boiling point (Tb) of the material: ")) # get user input for boiling point and convert to float
T = float(input("Enter the temperature (T) of the environment: ")) # get user input for temperature and convert to float
print("The state of the material at temperature T is: ", state(Tm, Tb, T)) # call the state function with the user input and print the result
````

## Sets
A set is an **unordered** collection of **unique** elements. 

Operations on sets include:
- `add(...)`: adds an element to the set
- `remove(...)`: removes an element from the set
- `in`: checks if an element is in the set
- `len(...)`: returns the number of elements in the set
- `intersection(...)`: returns a new set that is the intersection of two sets
- `union(...)`: returns a new set that is the union of two sets

To create a set, we can use curly braces `{}` or the `set(...)` constructor.

````python
s1 = {3,4,5} 
3 in s1
````

````python
s2 = set()        # empty set
s2.add(3)
s2.add(2)
s2
````

````python
s2 = {1, 2}
s2.add(2) # no effect, since 2 is already in
s2
````

````python
s2 = {1, 2}
s2.discard(2)
print("after discarding 2: ", s2)

s2.discard(5)
print("after discarding 5: ", s2) # no effect, since 5 is not in the set
````

````python
s1 = {1, 2, 3, 4, 5}
s2 = {3, 4, 5, 6, 7}
s3 = s1.union(s2) # union of s1 and s2
s4 = s1.intersection(s2) # intersection of s1 and s2

print("s1: ", s1)
print("s2: ", s2)
print("union of s1 and s2: ", s3)
print("intersection of s1 and s2: ", s4)
````

````python
# set can be used to remove duplicates from a list
lst = [1, 2, 2, 3, 4, 4, 5]
set_of_elements = set(lst) # convert list to set to remove duplicates
unique_lst = list(set_of_elements) # convert set back to list

print("original list: ", lst)
print("unique list: ", unique_lst)
````

## Mapping DNA codons to the amino acids using a dictionary

````python
# mapping of codons to amino acids 
# according to the standard genetic code
# `*` represents stop codons
universal = {
    'ttt': 'F', 'tct': 'S', 'tat': 'Y', 'tgt': 'C',
    'ttc': 'F', 'tcc': 'S', 'tac': 'Y', 'tgc': 'C',
    'tta': 'L', 'tca': 'S', 'taa': '*', 'tga': '*',
    'ttg': 'L', 'tcg': 'S', 'tag': '*', 'tgg': 'W',

    'ctt': 'L', 'cct': 'P', 'cat': 'H', 'cgt': 'R',
    'ctc': 'L', 'ccc': 'P', 'cac': 'H', 'cgc': 'R',
    'cta': 'L', 'cca': 'P', 'caa': 'Q', 'cga': 'R',
    'ctg': 'L', 'ccg': 'P', 'cag': 'Q', 'cgg': 'R',

    'att': 'I', 'act': 'T', 'aat': 'N', 'agt': 'S',
    'atc': 'I', 'acc': 'T', 'aac': 'N', 'agc': 'S',
    'ata': 'I', 'aca': 'T', 'aaa': 'K', 'aga': 'R',
    'atg': 'M', 'acg': 'T', 'aag': 'K', 'agg': 'R',

    'gtt': 'V', 'gct': 'A', 'gat': 'D', 'ggt': 'G',
    'gtc': 'V', 'gcc': 'A', 'gac': 'D', 'ggc': 'G',
    'gta': 'V', 'gca': 'A', 'gaa': 'E', 'gga': 'G',
    'gtg': 'V', 'gcg': 'A', 'gag': 'E', 'ggg': 'G'
}

````

````python
# translate a DNA sequence to a protein sequence using the universal genetic code
def dna_translate(gene, code=universal):
    """Translate a DNA sequence into a protein sequence."""

    prot = ""
    gene = gene.lower()  # match lowercase codons in the universal dictionary

    for i in range(0, len(gene), 3):
        codon = gene[i:i+3]
        if codon in code:
            prot += code[codon]      # append amino acid
        else:
            prot += "?"              # unknown or incomplete codon

    return prot

#Example usage:
dna_translate("ATGGACGGCGGCTAA") # should return "MDGG*"
````

# 1.11 Files 

### 📁 What Are Files?
- Used to store **long‑term data** beyond a program’s execution.

### 💾 Why Files Matter
- Can hold **much larger amounts of data** than variables in memory.
- Data **persists** even after:
  - the program ends  
  - the computer shuts down  

### 🧬 Files in Biology
- Essential for storing:
  - **genomes** of organisms  
  - **gene expression data** from healthy vs. mutant cells  

### 🖥️ How Files Are Used
- Programs later **read** these files to perform analyses and extract insights.

## Working with Files in Python
### 📂 Opening a File
- Use `open(filename, mode)` to open a file.
- `filename`: the name of the file (e.g., `"data.txt"`)
- `mode`: how you want to interact with the file:
  - `'r'`: read (default)
  - `'w'`: write (creates a new file or overwrites an existing one)
  - `'a'`: append (adds to the end of an existing file)

### 📖 Reading from a File
- Use `file.read()` to read the entire contents of the file as a string.
- Use `file.readline()` to read one line at a time.
- Use `file.readlines()` to read all lines into a list.

### ✍️ Writing to a File
- Use `file.write(string)` to write a string to the file.
- Use `file.writelines(list_of_strings)` to write a list of strings to the file.

### 🔒 Closing a File
- Always close a file after you're done using it with `file.close()`.




````python
# read from the file ./yeast_genome.txt
# translate the DNA sequence to a protein sequence using the universal genetic code
# and write the protein sequence to the file ./yeast_proteome.txt
my_file = open("../data/yeast_genome.txt", "r") # open the file for reading
# in Colab :
# Create "/content/data" directory if it doesn't exist
# !mkdir /content/data
# Upload "yeast_genome.txt"  and "yeast_proteome.txt" files to the "/content/data" directory

dna_sequence = my_file.read() # read the entire contents of the file into a string
my_file.close() # close the file

# remove any whitespace characters from the string
dna_sequence = dna_sequence.replace("\n", "") # remove newline characters

protein_sequence = dna_translate(dna_sequence) # translate the DNA sequence to a protein sequence
my_file = open("../data/yeast_proteome.txt", "w") # open the file for writing

my_file.write(protein_sequence) # write the protein sequence to the file
my_file.close() # close the file

# read the protein sequence from the file and print the first 100 characters
my_file = open("../data/yeast_proteome.txt", "r") # open the file for reading
protein_sequence = my_file.read() # read the entire contents of the file into a string
my_file.close() # close the file
print(protein_sequence[:100]) # print the first 100 characters of the protein sequence
````

# 1.12 Libraries 
### 📚 What Are Libraries?
- Collections of pre‑written code that provide additional functionality.
- Allow you to perform complex tasks without having to write all the code yourself.
### 🔍 Why Use Libraries?
- Save time and effort by reusing code that has already been tested and optimized.
- Provide access to a wide range of tools and functions for various tasks.

## 📖Examples of Libraries in Python
- **math**: for mathematical functions and constants.
- **random**: for generating random numbers and performing random operations.
- **time**: for working with time-related functions.

### 🧬 Libraries in Biology
- **Biopython**: for biological computations and data analysis.
- **Pandas**: for data manipulation and analysis.
- **Matplotlib**: for data visualization.
- **NumPy**: for numerical computations and handling large datasets.

````python
# Example: generate a random DNA sequence of length 50
import random # import the random module to generate random numbers
def random_dna(length):
    ''' generate a random DNA sequence of given length '''
    bases = ["A", "C", "G", "T"] # list of DNA bases
    dna = ""
    for i in range(length):
        dna += random.choice(bases) # randomly select a base and append to the dna string
    return dna

print(random_dna(50)) # generate and print a random DNA sequence of length 50
print(random_dna(50)) # generate and print another random DNA sequence of length 50
````

````python
# Example: generate a random permutation of numbers from 1 to n
def random_permutation(n):
    ''' generate a random permutation of numbers from 1 to n '''
    numbers = list(range(1, n+1)) # create a list of numbers from 1 to n
    random.shuffle(numbers) # shuffle the list in place to get a random permutation
    return numbers

print(random_permutation(10)) # generate and print a random permutation of numbers from 1 to 10
print(random_permutation(10)) # generate and print another random permutation of numbers from 1 to 10
````
